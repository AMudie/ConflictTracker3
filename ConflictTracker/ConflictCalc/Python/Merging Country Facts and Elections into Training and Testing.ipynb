# Merging Country Facts and Elections into Training and Testing.ipynb


```python
import pandas as pd
```

```python
df_country_facts = pd.read_csv("Country Data Preprocessed\Country Facts.csv")
df_country_facts["Year"] =  df_country_facts['Year'].astype(int)

#Drop data older than 2015, it's added by the World Bank Dataset tool autmatically. 
df_country_facts_rows = df_country_facts[df_country_facts["Year"] < 2015].index
df_country_facts.drop(df_country_facts_rows, inplace=True)

#Drop Time Code and Country Code columns: not required. 
df_country_facts = df_country_facts.drop(["Time Code", "Country Code"], axis=1)

df_country_facts.head()
```

```python
df_elections = pd.read_csv("Election Data original\Elections.csv")
df_elections["Date"] = pd.to_datetime(df_elections["Date"], format="%d/%m/%Y")
df_elections.head()
```

```python
df_context_ct_full = pd.read_csv("Dataset Preprocessed\df_train_chronos2_timeseries_preprocessed.csv")
df_context_ct_full["periodStart"] = pd.to_datetime(df_context_ct_full["periodStart"], format="%Y-%m-%d")
df_context_ct_full["periodEnd"] = pd.to_datetime(df_context_ct_full["periodEnd"], format="%Y-%m-%d")

df_context_ct_full.head()
```

```python
df_test_ct_full = pd.read_csv("Dataset Preprocessed\df_test_chronos2_timeseries_preprocessed.csv")
df_test_ct_full["periodStart"] = pd.to_datetime(df_test_ct_full["periodStart"], format="%Y-%m-%d")
df_test_ct_full["periodEnd"] = pd.to_datetime(df_test_ct_full["periodEnd"], format="%Y-%m-%d")
df_test_ct_full.head()
```

## Periods

```python
print("Train periods: ", df_context_ct_full["periodStart"].min(), " - " , df_context_ct_full["periodEnd"].max())
print("Test periods: ", df_test_ct_full["periodStart"].min(), " - " , df_test_ct_full["periodEnd"].max())
```

```python
## Before going any further, check the train and test sets do not contain NaNs (i.e. NaNs are being added by this sheet). 
```

```python
nan_rows_context = []

for col in df_context_ct_full.columns:
    # Identify rows where this column is NaN
    mask = df_context_ct_full[col].isna()

    if mask.any():
        subset = df_context_ct_full.loc[mask, ["country", "place","periodStart"]].copy()
        subset["column"] = col
        nan_rows_context.append(subset)

# Combine into one DataFrame
nan_context_df = pd.concat(nan_rows_context, ignore_index=True)

print(nan_context_df)
nan_context_df.to_csv("contextdata-NANs.csv")
```

## Add Election Flag

```python
def add_election_flag(df_periods, df_elections):
    df_periods = df_periods.copy()

    # Ensure periodStart and periodEnd are datetime
    df_periods["periodStart"] = pd.to_datetime(df_periods["periodStart"])
    df_periods["periodEnd"] = pd.to_datetime(df_periods["periodEnd"])

    # Add the flag column
    df_periods["election_flag"] = 0

    # Process country by country (fast and clean)
    for country in df_periods["country"].unique():
        # Election dates for this country
        election_dates = df_elections.loc[
            df_elections["Country"] == country, "Date"
        ]

        if election_dates.empty:
            continue  # no elections for this country

        # For each period, check if any election date falls inside the window
        mask = df_periods["country"] == country
        df_periods.loc[mask, "election_flag"] = df_periods.loc[mask].apply(
            lambda row: int(
                ((election_dates >= row["periodStart"]) &
                 (election_dates <= row["periodEnd"])).any()
            ),
            axis=1
        )

    return df_periods
```

```python
#Add the election flag column to the train and test datasets. 
df_context_ct_full = add_election_flag(df_context_ct_full, df_elections)
df_test_ct_full = add_election_flag(df_test_ct_full, df_elections)

df_test_ct_full
```

## Add the Country Facts features from df_country_facts to the train and testing datasets

```python
print(df_country_facts.columns)
```

```python
# Add the raw features (can add the imputation flags if required)
df_country_facts_feature_columns = [
    'Population, total [SP.POP.TOTL]',
    'Control of Corruption - Governance estimate (approx. -2.5 to +2.5) [GOV_WGI_CC_EST]',
    'Rule of Law - Governance estimate (approx. -2.5 to +2.5) [GOV_WGI_RL_EST]',
    'Primary completion rate, total (% of relevant age group) [SE.PRM.CMPT.ZS]',
    'Children out of school (% of primary school age) [SE.PRM.UNER.ZS]',
    'GNI per capita (constant 2015 US$) [NY.GNP.PCAP.KD]',
    'GDP per capita (constant 2015 US$) [NY.GDP.PCAP.KD]',
    'Armed forces personnel (% of total labor force) [MS.MIL.TOTL.TF.ZS]',
    'Military expenditure (% of GDP) [MS.MIL.XPND.GD.ZS]',
    'Renewable internal freshwater resources per capita (cubic meters) [ER.H2O.INTR.PC]',
    'Cereal yield (kg per hectare) [AG.YLD.CREL.KG]',
    'Permanent cropland (% of land area) [AG.LND.CROP.ZS]',
    'Water productivity, total (constant 2015 US$ GDP per cubic meter of total freshwater withdrawal) [ER.GDP.FWTL.M3.KD]',
    'Internally displaced persons, new displacement associated with conflict and violence (number of cases) [VC.IDP.NWCV]'
]

facts_subset = df_country_facts[["Country Name", "Year"] + df_country_facts_feature_columns]


#Context data:
df_context_ct_full["year"] = pd.to_datetime(df_context_ct_full["periodStart"]).dt.year
df_context_ct_full = df_context_ct_full.merge(
    facts_subset,
    left_on=["country", "year"],
    right_on=["Country Name", "Year"],
    how="left"
)
df_context_ct_full = df_context_ct_full.drop(["year"], axis=1)

#Testing data:
df_test_ct_full["year"] = pd.to_datetime(df_test_ct_full["periodStart"]).dt.year
df_test_ct_full = df_test_ct_full.merge(
    facts_subset,
    left_on=["country", "year"],
    right_on=["Country Name", "Year"],
    how="left"
)
df_test_ct_full = df_test_ct_full.drop(["year"], axis=1)
```

```python
nan_rows_context = []

for col in df_context_ct_full.columns:
    # Identify rows where this column is NaN
    mask = df_context_ct_full[col].isna()

    if mask.any():
        subset = df_context_ct_full.loc[mask, ["country", "place","periodStart"]].copy()
        subset["column"] = col
        nan_rows_context.append(subset)

# Combine into one DataFrame
nan_context_df = pd.concat(nan_rows_context, ignore_index=True)

print(nan_context_df)
nan_context_df.to_csv("contextdata-NANs.csv")
```

```python
nan_rows_test = []

for col in df_test_ct_full.columns:
    mask = df_test_ct_full[col].isna()

    if mask.any():
        subset = df_test_ct_full.loc[mask, ["country", "place", "periodStart"]].copy()
        subset["column"] = col
        nan_rows_test.append(subset)

nan_test_df = pd.concat(nan_rows_test, ignore_index=True)

nan_test_df.to_csv("testdata-NANs.csv")
```
